###### List

```python
countries = ['United States', 'India', 'China', 'Brazil']
```

```python
print(countries[0])
print(countries[1])
print(countries[2])
print(countries[3])
```

```python
print(countries[-1])
```

```python
# Slicing
print(countries[0:3])
print(countries[1:])
print(countries[:2])
```

```python
# Adding elements to a list
# append: Adds a new element at the end of a list
countries = ['United States', 'India', 'China', 'Brazil']
countries.append('Canada')
countries
```

```python
# insert: Adds a new element at a specific position
countries = ['United States', 'India', 'China', 'Brazil']
countries.insert(0,'Canada')
countries
```

```python
#Sorting a list
numbers = [4, 3, 10, 7, 1, 2]
numbers.sort()
numbers
```

```python
numbers = [4, 3, 10, 7, 1, 2]
numbers.sort(reverse=True)
numbers
```

```python
#Update value on a list
numbers[0] = 100
numbers
```

###### if - else - elif

```python
countries = ['United States', 'India',
             'China', 'Brazil']

if 'China' in countries:
    print("Country is in list")
else:
    print("Not in list")
```

```python
age = 18

if age>=18:
    print("You're an adult!")
elif age>=13:
    print("You're a teenager!")
else:
    print("You're a kid")
```

###### for loop

```python
for country in countries:
    print(country)
```

```python
for country in countries:
    if country == "United States":
        print(country)
```

```python
# Enumerating in a for loop
for i, country in enumerate(countries):
    print(i, country)
```
