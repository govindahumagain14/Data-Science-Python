```python
import pandas as pd
```

```python
# reading the csv file
df_population = pd.read_csv('population_total.csv')
```

```python
# first look at the dataset
df_population
```

##### Making a Pivot Table

```python
# dropping null values
df_population.dropna(inplace=True)
```

```python
# making a pivot table
df_pivot = df_population.pivot(index='year', columns='country', values='population')
```

```python
# selecting some countries
df_pivot = df_pivot[['United States', 'India', 'China', 'Indonesia', 'Brazil']]
```

```python
# showing pivot table
df_pivot
```

##### Lineplot

```python
df_pivot.plot(kind='line', xlabel='Years', ylabel='Population', title='Population (1955-2020)',  figsize=(8,4))
```

##### Barplot

###### Single Barplot

```python
# selecting only one year (2020)
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
# showing new dataframe
df_pivot_2020
```

```python
# transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot_2020.T
```

```python
# making barplot
df_pivot_2020.plot(kind='bar',color='red', xlabel='Years', ylabel='Population', title='Population in 2020')
```

##### Barplot grouped by “n” variables

```python
# selecting a few years
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
```

```python
# making grouped barplot
df_pivot_sample.plot(kind='bar', xlabel='Year', ylabel='Population', title='Population (1980-2020)', figsize=(8,4))
```

##### Piechart

```python
# changing column name 
df_pivot_2020.rename(columns={2020:'2020'}, inplace=True)
```

```python
# making the plot
df_pivot_2020.plot(kind='pie', y = '2020', title='Population in 2020 (%)', autopct='%1.1f%%')
```

##### Boxplot

###### Single Boxplot


```python
df_pivot['United States'].plot(kind='box', color='green', ylabel='Population')
```

###### Multiple Boxplot

```python
df_pivot.plot(kind='box', xlabel='Countries', ylabel='Population')
```

##### Histogram

```python
df_pivot[['United States', 'Indonesia']].plot(kind='hist', xlabel='Population')
                                              
```
